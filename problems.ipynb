```python
# 3.A Code
x = np.linspace(-1, 7, 500)
dist = stats.expon(scale=1)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(x, dist.cdf(x))
axes[0].set(title="Exponential CDF", xlabel="x", ylabel="F(x)")
axes[1].plot(x, dist.pdf(x))
axes[1].set(title="Exponential PDF", xlabel="x", ylabel="f(x)")
plt.tight_layout()
plt.show()
```

```python
# 4.A Code
x = np.linspace(-8, 8, 500)
dist = stats.logistic(loc=0, scale=1)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(x, dist.cdf(x))
axes[0].set(title="Logistic CDF", xlabel="x", ylabel="F(x)")
axes[1].plot(x, dist.pdf(x))
axes[1].set(title="Logistic PDF", xlabel="x", ylabel="f(x)")
plt.tight_layout()
plt.show()
```

5.C

Solving F(x) = u for x gives

$$
u=1-e^{-x}
\quad\Rightarrow\quad
e^{-x}=1-u
\quad\Rightarrow\quad
x=-\ln(1-u).
$$

The exponential quantile function is

$$
F_X^{-1}(u)=-\ln(1-u), \qquad 0<u<1.
$$

5.D

Solving F(x) = u for x gives

$$
u=\frac{1}{1+e^{-x}}
\quad\Rightarrow\quad
e^{-x}=\frac{1-u}{u}
\quad\Rightarrow\quad
x=\ln\left(\frac{u}{1-u}\right).
$$

The logistic quantile function is

$$
F_X^{-1}(u)=\ln\left(\frac{u}{1-u}\right), \qquad 0<u<1.
